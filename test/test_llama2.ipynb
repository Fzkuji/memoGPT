```python
from transformers import LlamaTokenizer, LlamaForCausalLM
import torch

# 加载模型和tokenizer
model_name = "meta-llama/Llama-2-7b-chat-hf"
tokenizer = LlamaTokenizer.from_pretrained(model_name)
model = LlamaForCausalLM.from_pretrained(model_name)


```

```python
for layer in model.model.layers:
    attention_layer = layer.self_attn
    print(attention_layer.rotary_emb.inv_freq.shape)
    break
```

```python
# 创建示例输入
input_text = "Hello, how are you?"
input_ids = tokenizer(input_text, return_tensors="pt").input_ids

```

```python
# 进行一次前向传播以提取 position_ids
with torch.no_grad():
    outputs = model(input_ids, output_hidden_states=True)
    # 在这个模型架构中，可能需要检查每层attention中的position_ids
    for layer in model.model.layers:
        attention_layer = layer.self_attn
        position_ids = attention_layer.forward(
            hidden_states=input_ids,
            position_ids=torch.arange(input_ids.size(1)).unsqueeze(0).expand(input_ids.size(0), -1)
        )  # position_ids 是在第三个返回值中
        # print(f"Layer {layer}: Position IDs - {position_ids}")

# 注意：这个实现假设position_ids在每个attention层中独立使用
# 如果你想要访问模型的不同attention层中的position_ids，请根据需要修改代码
```
